**Feature Engineering**

Since i already imputed the missing values with the entire dataset,if i go ahead and train that data, it will create a data leakage since the data is already exposed. Therefore, so to prevent this, i will relaod the original dataset.

**Split the Dataset to Prevent Data Leakage**

```python
# Separate features and target
X = df.drop(columns=["SalePrice", "Id"])
y = df["SalePrice"]

```

SalePrice is removed from X because it is the target varaible (what we are trying to predict). Id is also removed because it is primarily an identifier rather than a meaningful house characteristic.

```python
#80/20 split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
#perform the MinMax scaler on just the training dataset
from sklearn.preprocessing import MinMaxScaler

# Identify numerical features
numerical_features = X_train.select_dtypes(
    include=["int64", "float64"]
).columns

X_train_numeric = X_train[numerical_features]
X_test_numeric = X_test[numerical_features]

# Median imputation
num_imputer = SimpleImputer(strategy="median")

X_train_numeric = num_imputer.fit_transform(X_train_numeric)
X_test_numeric = num_imputer.transform(X_test_numeric)

# Min-Max scaling
scaler = MinMaxScaler()

X_train_scaled = scaler.fit_transform(X_train_numeric)
X_test_scaled = scaler.transform(X_test_numeric)
```

```python
#On the categorical variables
X_train_categorical = X_train[categorical_features]
X_test_categorical = X_test[categorical_features]

cat_imputer = SimpleImputer(strategy="most_frequent")

# Fit on training data
X_train_categorical = cat_imputer.fit_transform(
    X_train_categorical
)

# Transform test data
X_test_categorical = cat_imputer.transform(
    X_test_categorical
)
```

```python
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

# Fit encoder ONLY on training data
X_train_encoded = encoder.fit_transform(
    X_train_categorical
)

# Apply same encoder to test data
X_test_encoded = encoder.transform(
    X_test_categorical
)
```

To prevent data leakage, i reloaded the original dataset, then split the dataset into 80% training data and 20% testing data before fitting the preprocessing methods. Missing-value imputation, Min-Max scaling, and One-Hot Encoding should be fitted only on the training data. The fitted transformations are then applied to the test data without using information from the test set. This ensures that the model is evaluated on unseen data and provides a more reliable measure of its performance.

```python
import numpy as np
# Combine numerical and categorical training features
X_train_preprocessed = np.hstack(
    (X_train_scaled, X_train_encoded)
)

# Combine numerical and categorical test features
X_test_preprocessed = np.hstack(
    (X_test_scaled, X_test_encoded)
)

print("Training shape:", X_train_preprocessed.shape)
print("Testing shape:", X_test_preprocessed.shape)
```

**Train the Linear Regression model**

```python
from sklearn.linear_model import LinearRegression

# Create the model
model = LinearRegression()

# Train the model
model.fit(X_train_preprocessed, y_train)
```

```python
# Training predictions
y_train_pred = model.predict(X_train_preprocessed)

# Test predictions
y_test_pred = model.predict(X_test_preprocessed)
```

```python
#calculate the mse
from sklearn.metrics import mean_squared_error

# Calculate MSE
train_mse = mean_squared_error(
    y_train,
    y_train_pred
)

test_mse = mean_squared_error(
    y_test,
    y_test_pred
)

print("Training MSE:", train_mse)
print("Test MSE:", test_mse)
```

The Linear Regression model achieved a training MSE of 381,425,292.45 and a test MSE of 868,806,561.31. The higher test MSE indicates that the model performs less accurately on unseen data and suggests some overfitting

```python
#Calculate training and testing RMSE
train_rmse = np.sqrt(train_mse)
test_rmse = np.sqrt(test_mse)

print("Training RMSE:", train_rmse)
print("Test RMSE:", test_rmse)
```

The Linear Regression model achieved a training RMSE of $19,530.11 and a test RMSE of $29,475.59. The higher test RMSE shows that the model performs better on the training data than on unseen data. This suggests some overfitting. $29,475.59 on the test data indicates that the model's predicted house sale prices typically deviate from the actual SalePrice by $29,476, with larger prediction errors being penalized more heavily.
