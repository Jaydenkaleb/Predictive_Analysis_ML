```python
#Create a dataset in CSV format containing at least 100 records of fictional user data (Name, Age, City)
names = [
    "John", "Sarah", "Michael", "Emily", "David",
    "Jessica", "Daniel", "Sophia", "James", "Olivia"
]

cities = [
    "Toronto", "Vancouver", "Calgary", "Ottawa", "Montreal",
    "Halifax", "Winnipeg", "Edmonton", "St. John's", "Victoria"
]
```

```python
#Automatically generate 100 fictional records
data = []

for i in range(100):
    record = {
        "Name": random.choice(names),
        "Age": random.randint(10, 80),
        "City": random.choice(cities)
    }
    
    data.append(record)

df = pd.DataFrame(data)

print(df.head())
print("\nNumber of records:", len(df))
```

```python
#convert that into a CSV file
df.to_csv('samplefictional_data.csv', index=False)
print('csv file created successfully')
```

```python
#Coverting to parguet
df.to_parquet("samplefictional_data.parquet", index=False)

print("Parquet file created successfully.")
```

```python
#compare the file sized
csv_size = os.path.getsize("samplefictional_data.csv")
parquet_size = os.path.getsize("samplefictional_data.parquet")

print("CSV file size:", csv_size, "bytes")
print("Parquet file size:", parquet_size, "bytes")

```

Note that because the file size is small. Imformation shows that the with only 100 small records, the Parquet file is actually larger than the CSV file because, parquet contains metadata and has overhead that may outweigh its compression benefits for very small datasets.

```python
#Transform the data by filtering out records where the age is below 18 and converting the names to uppercase.
df_transform = df_extract[df_extract["Age"] >= 18].copy()

df_transform["Name"] = df_transform["Name"].str.upper()

print(df_transform.head())
```

```python
#Load the transformed data into a new CSV file.
df_transform.to_csv("transformed_users.csv", index=False)

print("Transformed data saved successfully.")
```

```python
#Verify the transformation by loading and displaying the new file.
verified_df = pd.read_csv("transformed_users.csv")

print(verified_df.head())
print("\nNumber of transformed records:", len(verified_df))
```

The users below 18 years old were removed, and all user names were converted to uppercase. The transformed data was then loaded into a new CSV file named transformed_users.csv. Finally, the new file was loaded into a DataFrame to verify that the transformations were successfully applied.

**Exploring Data Models with SQLite**

```python
#import sqlite
import sqlite3
```

```python
#Create/connect to a database
connection = sqlite3.connect("users.db")

cursor = connection.cursor()

print("Connected to SQLite successfully.")
```

```python
#Create the users table
cursor.execute("""
CREATE TABLE IF NOT EXISTS users (
    ID INTEGER PRIMARY KEY,
    Name TEXT,
    Age INTEGER,
    City TEXT
)
""")

connection.commit()

print("Users table created successfully.")
```

```python
#Insert at least 5 records into the table and retrieve the data using an SQL query in python.
users = [
    (1, "John", 25, "Toronto"),
    (2, "Sarah", 32, "Vancouver"),
    (3, "Michael", 28, "Calgary"),
    (4, "Emily", 40, "Ottawa"),
    (5, "David", 35, "Montreal")
]

cursor.executemany("""
INSERT OR REPLACE INTO users (ID, Name, Age, City)
VALUES (?, ?, ?, ?)
""", users)

connection.commit()

print("Records inserted successfully.")
```

```python
#Retrieve the records using SQL.
cursor.execute("SELECT * FROM users")

results = cursor.fetchall()

for row in results:
    print(row)
```

```python
#test mongoDb has been install
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017/")

db = client["UserDatabase"]
collection = db["users"]

print("Connected to MongoDB successfully.")
```

```python
#Create five documents in MongoDB. Note that mongo stores rows as documents
mongo_users = [
    {"ID": 1, "Name": "John", "Age": 25, "City": "Toronto"},
    {"ID": 2, "Name": "Sarah", "Age": 32, "City": "Vancouver"},
    {"ID": 3, "Name": "Michael", "Age": 28, "City": "Calgary"},
    {"ID": 4, "Name": "Emily", "Age": 40, "City": "Ottawa"},
    {"ID": 5, "Name": "David", "Age": 35, "City": "Montreal"}
]
```

```python
#Insert the records as documents and retrieve them using a MongoDB query
collection.delete_many({})

collection.insert_many(mongo_users)

print("MongoDB records inserted successfully.")

#retrieve
results = collection.find()

for document in results:
    print(document)
```

Comparison of SQLite and MongoDB Data Models
SQLite and MongoDB use different approaches to storing data. SQLite uses a relational model, where data is organized into tables consisting of predefined rows and columns. SQL queries such as SELECT are used to retrieve the stored data. This structured approach makes SQLite suitable for data with clearly defined relationships and consistent fields.
MongoDB uses a document model, where records are stored as flexible JSON-like documents. Unlike a relational table, documents in the same MongoDB collection do not necessarily need to contain exactly the same fields. This makes MongoDB more flexible when working with data whose structure may change.
SQLite was straightforward to use because Python provides built-in support through sqlite3, and its table structure makes the data easy to organize and query. MongoDB required additional setup and the pymongo library, but its document structure provided greater flexibility.
Overall, SQLite is well suited for structured relational data where consistency and relationships are important, while MongoDB is useful when working with flexible, semi-structured, or evolving data.
